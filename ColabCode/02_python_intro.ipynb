### 三角関数、平方根、指数、対数の使い方

## 3. 繰り返し処理
ここでは、`for` 文による繰り返し処理を扱います。
物理シミュレーションでは、微小時間ステップごとに更新計算を行うため、繰り返しは最重要の構文です。

### Python と C の対応
- Python: `for i in range(1, 6):`
- C: `for (int i = 1; i < 6; i++) { ... }`

主な違いは次の3点です。
1. Pythonは `range` で反復範囲を作る（終端値は含まない）。
2. Pythonは `:` とインデントでブロックを表す。Cは `{}` を使う。
3. Pythonは変数宣言が不要。Cは `int i` のように型宣言が必要。

```python
# 例: 1から5までの合計を計算するループ
# Cで同じ処理を書くと:
# int total = 0;
# for (int i = 1; i < 6; i++) {
#     total += i;
#     printf("i = %d までの合計 = %d\n", i, total);
# }

total = 0  # Pythonは型宣言なし（Cなら int total = 0;）
for i in range(1, 6):  # range(1, 6) は 1,2,3,4,5（6は含まない）
    total += i
    print("i =", i, "までの合計 =", total)  # Cなら printf を使う
```

<summary>種明かし...演習を解いて考察したあとに見てください</summary>

上の演習では，真値とわずかなずれが生じたはずです．
これは **丸め誤差（round-off error）** と呼ばれ，数を有限の桁数で表現することによって起こります．
具体的には，倍精度（64bit）の `float` は約16桁，単精度（32bit）の `float32` は約7桁の有効桁数で実数を表します．
観察した誤差のオーダーは，この表現精度と対応していましたね．
なお，表せる実数の大きさの範囲はそれぞれ，倍精度：$1.7 \times 10^{-308}$ ～ $1.7 \times 10^{308}$，単精度：$3.4 \times 10^{-38}$ ～ $3.4 \times 10^{38}$程度です．


<div style="border-left: 5px solid #3b82f6; background-color: #eff6ff; padding: 12px; border-radius: 6px;">
<b>📝 NOTE　単精度と倍精度とは</b><br>
コンピュータでは，数を2進数（0と1）で表します．整数の場合は比較的簡単で，例えば2進数 `11011` は，
`1×2^4 + 1×2^3 + 0×2^2 + 1×2^1 + 1×2^0 = 16 + 8 + 0 + 2 + 1 = 27`
となり，10進数の `27` を表します．

この 0 と 1 の1桁を **ビット** と呼びます．ビット数が多いほど表現できる情報量は増えますが，ビット数が有限である以上，表せる数の範囲や精度には限界があります．

実数（浮動小数点数）も同様に有限個のビットで表すため，丸め誤差が生じます．
一般に，単精度（float32）は有効桁数が約7桁，倍精度（float64）は約16桁です．
そのため，単精度のほうが誤差が現れやすくなります．
※ 浮動小数点数の詳細な内部表現（符号部・指数部・仮数部）は，別の授業（プログラミングII）で扱います．

### 4.3 情報落ちと桁落ちの課題

**情報落ち**

他にも誤差が起きる状況について演習してみましょう．

```python
import numpy as np # NumPyライブラリをインポート
a = np.float32(1) # NumPyのfloat32型変数を定義して，計算精度を単精度にします．
b = np.float32(123456700) 
c = np.float32(0)
c = a + b
print("a = ", a, ", b =", b, ", c = ", c)
```

上記のプログラムを解読して，何をやっているかを理解し，誤差が出てきているところ（おかしい結果）を見つけてください．
そして，その原因を考えてみましょう．

<details>
<summary>種明かし...演習を解いて考察したあとに見てください</summary>

**観察された現象**

単精度で定義した $a = 1$，$b = 123456700$ を足し算すると，
本来 $c = 123456701$ となるはずが，**$c = 123456700$** になってしまいます．
つまり，足し算をしたのに $a$ の値が消えてしまった状況です．

**原因の説明**

これは **情報落ち** と呼ばれる丸め誤差の一種です．

単精度（32bit）の `float32` は約7桁の有効桁数です．
個別に見ると：
- $a = 1$（1桁）
- $b = 1.234567 \times 10^8$（7桁）

どちらも `float32` で表現可能ですが，問題は「足し算」にあります．

$a + b = 1.23456701 \times 10^8$ という結果を表現しようとすると，
全体で8桁以上の精度が必要になります．
ところが `float32` の有効桁数は約7桁なので，
小さい桁の情報（この場合は 1）は丸め処理で失われてしまいます．

結果として，小さい値が大きい値に「吸収」され，計算したはずの小さい値が失われてしまうのです．
これを **「情報落ち」（loss of significance）** と呼びます．

</details>

$$a x^2 + b x + c = 0$$
の解を2通りの方法で求めます．

1つ目は，よく知られている**解の公式**
$$x_1 = \frac{-b + \sqrt{b^2 - 4ac}}{2a}$$
$$x_2 = \frac{-b - \sqrt{b^2 - 4ac}}{2a}$$
をそのまま使う方法です．

2つ目の方法は，一方の解を解の公式で求め，
$$x_2 = \frac{-b - \sqrt{b^2 - 4ac}}{2a}$$
もう一方の解は
$$x_1 = \frac{c}{a} / x_2$$
を使って求めてみるやり方です．．

このあと，プログラムの穴埋めをしながら，
- 解の公式をそのまま使った結果
- 片方の解を使ってもう一方を求めた結果
を比較し，どちらが解の精度が良いかを確かめてください．

```python
import numpy as np # NumPyライブラリをインポート
# 二次方程式 a x^2 + b x + c = 0 の桁落ちの例
# b が非常に大きい場合、解の公式の一方で精度を失いやすい
a = 1.
b = 1.e5 # eは10のべき乗を表す指数表記で、1.e5 は 1 × 10^5 = 100000 を意味します。
c = 1.

D = b**2 - 4*a*c
sqrtD = np.sqrt(D)

# 1) 普通に解の公式で2つとも計算
x1_naive = (-b + sqrtD) / (2*a)
x2_naive = (-b - sqrtD) / (2*a)

# 2) 桁落ちしにくい方を公式で計算し、もう一方は x1*x2=c/a から求める
# b > 0 のときは (-b - sqrtD)/(2a) 側が安定
x2_stable = (-b - sqrtD) / (2*a)
x1_stable = 0 # TODO 　

print("係数: a =", a, ", b =", b, ", c =", c)
print("\n[普通の解の公式]")
print("x1_naive =", x1_naive)
print("x2_naive =", x2_naive)

print("\n[片方を x1*x2=c/a で計算]")
print("x1_stable =", x1_stable)
print("x2_stable =", x2_stable)

# 検算: 方程式に代入したときの残差
r1_naive = a*x1_naive**2 + b*x1_naive + c
r2_naive = a*x2_naive**2 + b*x2_naive + c
r1_stable = a*x1_stable**2 + b*x1_stable + c
r2_stable = a*x2_stable**2 + b*x2_stable + c

print("\n[残差 ax^2+bx+c]")
print("naive の残差:", r1_naive, r2_naive)
print("stable の残差:", r1_stable, r2_stable)
```

**観察された現象**

上のプログラムを実行すると，2つの方法で得られた $x_1$ の値が大きく異なります．
特に「普通の解の公式」で計算した場合，残差 $ax^2+bx+c$ が `naive` では非常に大きくなるのに対し，
「修正版」の `stable` ではほぼゼロになっていることが分かります．

**原因：桁落ち（cancellation error）**

解の公式では
$$x_{1,2} = \frac{-b \pm \sqrt{b^2-4ac}}{2a}$$

この例では $b = 10^5$，$a = 1$，$c = 1$ なので $b^2 - 4ac \approx b^2$ です．
したがって $\sqrt{b^2-4ac} \approx b$ となり，ほぼ同じ大きさになります．

「普通の解の公式」では $x_1 = \frac{-b + \sqrt{b^2-4ac}}{2a}$ を計算しますが，
$b$ と $\sqrt{b^2-4ac}$ がほぼ同じ値であるため，その差を取ることになります．
これは非常に大きな値（$\sim 10^5$）から，それに非常に近い別の値を引く計算です．

浮動小数点数の有効桁数は限られているため，
このような「大きな数どうしの差」を取ると，わずかな丸め誤差が相対的に大きく拡大されてしまいます．
この現象を **桁落ち（cancellation error）** と呼びます．

**解決策：式変形**

そこで，より安定な計算方法を使います．
解の積の公式 $x_1 x_2 = \frac{c}{a}$ を利用すれば，
精度が高い方（$x_2 = \frac{-b - \sqrt{b^2-4ac}}{2a}$）をまず計算してから，
$x_1 = \frac{c}{a x_2}$ として求めることができます．

こうすることで，「大きな数どうしの差」という危険な計算を避けられ，
桁落ちによる誤差を大幅に軽減できるのです．

## 5. NumPyによるベクトル・行列演算（余裕がある人向けコンテンツ（授業では扱わない））
シミュレーションでは、複数の値をまとめて扱うために配列や行列を使います。`NumPy` を使うと、ベクトルや行列の計算を簡潔に書けます。

```python
import numpy as np

A = np.array([[1.0, 2.0], [3.0, 4.0]])
B = np.array([[2.0, 0.0], [1.0, 2.0]])
x = np.array([1.0, -1.0])

print("A =\n", A)
print("B =\n", B)
print("x =", x)
print("A + B =\n", A + B)
print("A @ x =", A @ x)
print("A @ B =\n", A @ B)

# 練習: 行列やベクトルの値を書き換えて結果を確認する
```

### このセルで行っている演算の意味
各行で行っている計算を順に説明します。

1. `A = np.array([[1.0, 2.0], [3.0, 4.0]])`
   2 x 2 の行列を作っています。
   $$A = \begin{bmatrix}1 & 2 \\ 3 & 4\end{bmatrix}$$

2. `B = np.array([[2.0, 0.0], [1.0, 2.0]])`
   もう1つの 2 x 2 行列です。
   $$B = \begin{bmatrix}2 & 0 \\ 1 & 2\end{bmatrix}$$

3. `x = np.array([1.0, -1.0])`
   2次元の列ベクトル（NumPy上は1次元配列）を作っています。
   $$x = \begin{bmatrix}1 \\ -1\end{bmatrix}$$

4. `A + B`
   行列の**要素ごとの足し算**です。同じ位置の成分同士を足します。
   $$A+B=\begin{bmatrix}1+2 & 2+0 \\ 3+1 & 4+2\end{bmatrix}=\begin{bmatrix}3 & 2 \\ 4 & 6\end{bmatrix}$$

5. `A @ x`
   行列とベクトルの積です。各行とベクトルの内積を計算します。
   $$A x = \begin{bmatrix}1\cdot1 + 2\cdot(-1) \\ 3\cdot1 + 4\cdot(-1)\end{bmatrix}=\begin{bmatrix}-1 \\ -1\end{bmatrix}$$

6. `A @ B`
   行列どうしの積です。左行列の行と右行列の列の内積で各要素が決まります。
   $$AB=\begin{bmatrix}1\cdot2+2\cdot1 & 1\cdot0+2\cdot2 \\ 3\cdot2+4\cdot1 & 3\cdot0+4\cdot2\end{bmatrix}=\begin{bmatrix}4 & 4 \\ 10 & 8\end{bmatrix}$$

ポイント: `+` は要素ごとの演算、`@` は線形代数の積です。

### 逆行列・固有値・対角化
行列の計算では、次の3つが特に重要です。

1. 逆行列
   行列 $A$ に対して $A^{-1}$ が存在するとき、
   $$A A^{-1} = A^{-1}A = I$$
   を満たします。連立一次方程式 $Ax=b$ の解は $x=A^{-1}b$ と書けます。

2. 固有値と固有ベクトル
   $$A v = \lambda v$$
   を満たす $\lambda$ を固有値、$v$ を固有ベクトルといいます。
   行列がどの方向をどれだけ伸縮するかを表す量です。

3. 対角化
   固有ベクトルを並べた行列を $P$、固有値を対角に並べた行列を $D$ とすると、
   $$A = P D P^{-1}$$
   と表せる場合があります。これを対角化といい，行列の性質を知るためにはとても重要な手法でした．

これらは，Pythonでライブラリを使えば比較的簡単に計算することができます．
やってみましょう．

```python
# 逆行列・固有値・対角化の例
A2 = np.array([[4.0, 1.0], [2.0, 3.0]])
print("A2 =\n", A2)

# 1) 逆行列
A2_inv = np.linalg.inv(A2)
print("A2^{-1} =\n", A2_inv)
print("A2 @ A2^{-1} =\n", A2 @ A2_inv)

# 2) 固有値・固有ベクトル
eigvals, eigvecs = np.linalg.eig(A2)
print("固有値 =", eigvals)
print("固有ベクトル（列ごと） =\n", eigvecs)

# 3) 対角化の確認: A2 = P D P^{-1}
P = eigvecs
D = np.diag(eigvals)
P_inv = np.linalg.inv(P)
A2_reconstructed = P @ D @ P_inv
print("P D P^{-1} =\n", A2_reconstructed)
print("再構成誤差の最大値 =", np.max(np.abs(A2 - A2_reconstructed)))
```
