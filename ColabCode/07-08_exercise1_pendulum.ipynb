## 相図による振り子の運動の理解

状態空間で運動の軌跡を表したものを相図と呼びました．
今回の例では，横軸に$\theta$，縦軸に$\omega$を取った相図を書くことができます．
相図を見ることで，実際どのような運動が起きているかを確認することができます．

<div style="border: 2px solid #000000; padding: 12px; border-radius: 8px; background: #f7dada;">
<b>演習問題 (7) </b><br>
ルンゲクッタ法のプログラム($dt = 0.01$とし，それ以外の物理パラメタは以前と同じ)を用いて，さまざまな初期値からはじめたときの運動の軌跡（相図）を書いてみましょう．<br>
さらに，代表的な軌跡（少なくとも下記の初期値６つについては説明すること）について，振り子のどのような運動に対応しているかを，振り子の絵をかきながら説明してください．
</div>
具体的には，以下の初期値については必ずプロットしてください．
$(\theta, \omega) = (0, 0), (\pi/2, 0), (\pi, 0), (-\pi, 0), (-\pi, 1), (\pi, -1)$
それ以外のプロットも歓迎いたします．

また，$\theta$が周期変数($-\pi$ ～ $\pi$)に注意しながらグラフを解釈してください．

```python
import numpy as np
import matplotlib.pyplot as plt

# ルンゲクッタ法の関数を実装できていれば，そのまま動くはずです．
#  4次のルンゲクッタ法で単振り子を解く関数
#  def runge_kutta_pendulum(theta0, omega0, g, l, dt, T):

# ここから具体的な条件を設定する
g = 9.8          # 重力加速度 [m/s^2]
l = 1.0          # 振り子の長さ [m]
dt = 0.01        # 時間刻み幅 [s]
T = 20.0         # 総時間 [s]

# 必須の初期値
initial_conditions = [
    (0.0, 0.0),
    (np.pi / 2.0, 0.0),
    (np.pi, 0.0),
    (-np.pi, 0.0),
    (-np.pi, 1.0),
    (np.pi, -1.0),
]

plt.figure(figsize=(9, 6))

for theta0, omega0 in initial_conditions:
    t, theta, omega = runge_kutta_pendulum(theta0, omega0, g, l, dt, T)
    line, = plt.plot(theta, omega, label=f"theta0={theta0:.2f}, omega0={omega0:.2f}")
    plt.plot(theta[0], omega[0], "o", markersize=5, color=line.get_color())

# サドル点の位置が見やすいように基準線を引く
plt.axvline(np.pi, color="gray", linestyle="--", linewidth=1)
plt.axvline(-np.pi, color="gray", linestyle="--", linewidth=1)
plt.axhline(0.0, color="gray", linestyle="--", linewidth=1)

plt.xlim(-1.1 * np.pi, 1.1 * np.pi)
plt.xlabel("theta [rad]")
plt.ylabel("omega [rad/s]")
plt.title("Phase portrait of the pendulum")
plt.grid(True)
plt.legend()
plt.show()

print("計算に使用した dt =", dt)
print("計算に使用した初期値 =", initial_conditions)
```
