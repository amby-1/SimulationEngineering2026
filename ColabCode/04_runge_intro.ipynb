## 1. ホイン法（2次のルンゲクッタ法）を実装してみる

最初に、解析解が分かっている最も基本的な例でホイン法を試します。
考える方程式は
$$
\frac{dx}{dt}=kx
$$
です。初期値を $x(0)=x_0$ とすると、真値は
$$
x(t)=x_0 e^{kt}
$$
となります。

ここでは、$x_0=1.0$, $k=-9.0$, $T=1.0$ を使います。

### 1.0 固定刻み幅で 1 ステップずつ計算してみる
時間刻み $\Delta t = 0.1$ として、ホイン法を実装してみましょう。
まずは 1 ステップずつ手で書いて、更新の形に慣れてみます。

ホイン法では、
1. まずオイラー法で仮の次の値を作る
2. その仮の値も使って傾きを平均する
という流れで次の値を計算します。

具体的には，
$$
k_1 = \Delta t f(t_n, x_n), \qquad k_2 = \Delta t f(t_n + \Delta t, x_n +  k_1)
$$
として、
$$
x_{n+1} = x_n + \frac{1}{2}(k_1 + k_2)
$$
で更新しましたね．

```python
import numpy as np
import matplotlib.pyplot as plt


def exact_linear(t, k, x0):
    return x0 * np.exp(k * t)


x0 = 1.0
k = -9.0
T = 1.0
dt = 0.1

t = np.linspace(0.0, T, 11)
x = np.full(11, np.nan)
x[0] = x0

# TODO: 下の 10 行を完成させてください
# ヒント: オイラー法を参考にして、ホイン法の更新式を使って x[1],...,x[10] を計算してください

t_true = np.linspace(0.0, T, 400)
x_true = exact_linear(t_true, k, x0)

plt.figure(figsize=(7, 4))
plt.plot(t_true, x_true, label='exact', linewidth=2)
plt.plot(t, x, 'o-', label=f'Heun dt={dt}')
plt.xlabel('t')
plt.ylabel('x')
plt.title('Time evolution of dx/dt = kx (Heun method)')
plt.grid(True)
plt.legend()
plt.show()

print('1秒後の真値 =', exact_linear(T, k, x0))
print('1秒後のホイン法 =', x[-1])
```

## 2. 4次のルンゲクッタ法を実装してみる

次は、1章と同じ微分方程式
$$
\frac{dx}{dt}=kx
$$
を、4次のルンゲクッタ法で解いてみます。

4次のルンゲクッタ法では
$$
k_1 = \Delta t f(t_n, x_n),\quad
k_2 = \Delta t f\left(t_n + \frac{\Delta t}{2}, x_n + \frac{1}{2}k_1\right),\quad
k_3 = \Delta t f\left(t_n + \frac{\Delta t}{2}, x_n + \frac{1}{2}k_2\right),\quad
k_4 = \Delta t f(t_n + \Delta t, x_n + k_3)
$$
を計算して、
$$
x_{n+1} = x_n + \frac{1}{6}(k_1 + 2k_2 + 2k_3 + k_4)
$$
で更新します。
