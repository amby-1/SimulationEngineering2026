## 1. 簡単な1次微分方程式でオイラー法を実装

最初に、解析解が分かっている最も基本的な例でオイラー法を試します。

考える方程式は
$$
\frac{dx}{dt}=kx
$$
です。初期値を $x(0)=x_0$ とすると、真値は
$$
x(t)=x_0 e^{kt}
$$
となります。

## 2. 同じ問題を陰解法オイラー法で解いてみる

つぎは、1章とまったく同じ微分方程式
$$
\frac{dx}{dt}=kx
$$
を、**陰解法のオイラー法**で解いてみます。

陽解法では
$$
x_{n+1}=x_n + \Delta t\,k x_n
$$
でしたが、陰解法では右辺にも次の時刻の値 $x_{n+1}$ が入ります。
つまり、更新式は
$$
x_{n+1}=x_n + \Delta t\,k x_{n+1}
$$
です。
この式を $x_{n+1}$ について解くと、
$$
x_{n+1}=\frac{x_n}{1-k\Delta t}
$$
となります。

1章では、刻み幅を粗くすると発散しました。
ここでは、陰解法にすると挙動がどう変わるかを、1章と同じ順番で確かめていきましょう。
